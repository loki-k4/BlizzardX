```python
# forecast_visuals.ipynb

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set(style="whitegrid")

# Load the forecast data
forecast_path = '/workspaces/BlizzardX/Data/forecast_20250425.csv'
df_forecast = pd.read_csv(forecast_path)

print(f"Loaded forecast data with {df_forecast.shape[0]} rows.")
df_forecast.head()
```

```python
# Choose a sample station
sample_station = df_forecast['Station_ID'].unique()[0]  # Pick the first station
print(f"Plotting for Station: {sample_station}")

df_station = df_forecast[df_forecast['Station_ID'] == sample_station]

plt.figure(figsize=(10, 6))
sns.lineplot(x='DATE', y='Predicted_TMIN', data=df_station, marker="o")
plt.title(f"7-Day Predicted TMIN for {sample_station}")
plt.xlabel("Date")
plt.ylabel("Predicted Minimum Temperature (°C)")
plt.xticks(rotation=45)
plt.show()

```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='DATE', y='Predicted_SNOW', data=df_station)
plt.title(f"7-Day Predicted Snowfall for {sample_station}")
plt.xlabel("Date")
plt.ylabel("Predicted Snowfall (cm)")
plt.xticks(rotation=45)
plt.show()
```

```python
cold_events = df_forecast[df_forecast['Predicted_Cold_Event'] == 1]
print(f"Number of extreme cold event predictions: {cold_events.shape[0]}")

cold_events.groupby('Station_ID').size().sort_values(ascending=False).head(10)
```
